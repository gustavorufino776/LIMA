### Laço FOR

```python
# Laço para interar
for i in range(1, 11):
  print(i)
```

```python
#Interar em listas

frutas = ['banana', 'abacaxi', 'laranja']

for fruta in frutas:
  print(fruta)
```

###Laço WHILE

```python
contador = 0
while contador < 10:
  print(contador)
  contador += 1
```

```python
opcao = 'sair'

while opcao != 'sair':
  print('Dentro do laço WHILE')
  opcao = input('digite \'sair\' para sair:')
```

```python
opcao = 'sair'

while True:
  print('Dentro do laço WHILE')
  opcao = input('digite \'sair\' para sair:')

  if opcao == 'sair':
    break
```

- Se escolher a opção 1, deve ser impresso "Exibindo perfil".
- Se escolher a opção 2, deve ser impresso "Alterando senha".
- O menu deve reaparecer sempre que o usuário escolher 1 ou 2.
- Se digitar 3, deve imprimir "Saindo do sistemas..." e encerrar.
- se digitar qualquer outra opção, deve imprimir "Opção invalida".

```python
#Desafio 2
soma_total = 0

while True:
  numero = int(input('digite um número: '))
  if numero <= 1:
    print(f'soma_total: {soma_total}')
    break
  elif numero > -1:
    soma_total += numero
```

### Funções

```python
#posíveis resutados de uma função:
#1) retornar "alguma coisa"; 2) causar algun efeito

def soma_dois_numeros(num1: int, num2: int) -> int: #type hinting
  soma = num1 + num2
  print(f'Soma dentro da função: {soma} (efeito colateral)') # efeito colateral
  return soma # retorna "alguma coisa"
  print("Depois do return...") # Nada é executado depois do return

soma_dois_numeros(1, 1) + 5
```

```python
# Estrutura try-except

def soma_dois_numeros(num1: int, num2: int) -> int: #type hinting
  try:
    return int(num1) + int(num2) #early return
  except TypeError:
    print('Os argumentos devem ser do tipos inteiros')
  except:
    print('aconteceu um erro, verifique suas entradas.')

try:
  print(soma_dois_numeros(10, 'dois') + 5)
except(TypeError, ValueError):
  print('Os argumentos devem ser do tipo inteiro e não podem ser palavras.')
```

Crie um programa com uma função que leia o nome de uma pessoa e imprima uma mensagem dizendo quantos anos a pessoa fará no próximo ano.(crie e chame uma função para isso, tratando erros de digitação com exeções).

```python
def ano(nome: str, idade: int):
  while True:
    try:
      nome = input('digite seu nome: ')
      idade = int(input('digite sua idade: '))
      if idade < 0:
        print('digite uma idade válida')
      idade += 1
      print(f"{nome}, você fara {idade}")
    except ValueError:
      print('Sua idade deve estar errada \ncoloque um número inteiro')

print(ano(nome = '', idade = 0))
```
